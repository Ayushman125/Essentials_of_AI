#Implement k_means clustering from scratch

###Question) Write a Python script to implement K-Means Clustering, a centroid-based clustering technique.

```python
import numpy as np

def kmeans_from_scratch(X, k, initial_centroids, max_iters=100, tol=1e-4):
    """
    Implements k-Means clustering from scratch.

    Parameters:
    - X (np.ndarray): Input data of shape (n_samples, n_features).
    - k (int): Number of clusters.
    - initial_centroids (np.ndarray): Initial cluster centers of shape (k, n_features).
    - max_iters (int): Maximum number of iterations.
    - tol (float): Convergence tolerance for centroid movement.

    Returns:
    - centroids (np.ndarray): Final computed centroids.
    - labels (np.ndarray): Cluster assignments for each data point.
    """
    centroids = np.array(initial_centroids, dtype=float)
    n_samples = X.shape[0]
    labels = np.zeros(n_samples, dtype=int)

    for iteration in range(max_iters):
        # Step 1: Assignment Step
        # Compute Euclidean distance from each data point to each centroid
        # Shape of distances: (n_samples, k)
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)

        # Assign each point to the closest centroid
        new_labels = np.argmin(distances, axis=1)

        # Step 2: Check for convergence (if assignments stop changing)
        if np.all(labels == new_labels):
            print(f"Converged: Labels unchanged at iteration {iteration}.")
            break

        labels = new_labels

        # Step 3: Update Step
        # Recalculate centroids as the mean of all points assigned to the cluster
        new_centroids = np.zeros_like(centroids)
        for i in range(k):
            cluster_points = X[labels == i]
            if len(cluster_points) > 0:
                new_centroids[i] = np.mean(cluster_points, axis=0)
            else:
                # Handle empty cluster edge case by retaining the old centroid
                new_centroids[i] = centroids[i]

        # Check convergence based on centroid shift distance
        shift = np.linalg.norm(new_centroids - centroids)
        if shift < tol:
            print(f"Converged: Centroid shift below tolerance at iteration {iteration}.")
            centroids = new_centroids
            break

        centroids = new_centroids

    return centroids, labels
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# 1. Generate synthetic data
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=42)

# 2. Define parameters
k = 3
# Manually choose or randomly sample initial centroids from the input data
np.random.seed(42)
initial_indices = np.random.choice(X.shape[0], k, replace=False)
initial_centroids = X[initial_indices]

print("Initial Centroids:\n", initial_centroids)

# 3. Run k-Means
final_centroids, labels = kmeans_from_scratch(X, k, initial_centroids)

print("\nFinal Centroids:\n", final_centroids)
```
