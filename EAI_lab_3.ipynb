## Q1. Perform the analysis of MOS on sample data point sets
[range,variance and standard deviation]

```python
import random
import numpy as np
data = [random.randint(1,100) for i in range(1,1000)]

def analyze_data(data):
    range = max(data) - min(data)
    variance = np.var(data)
    standard_deviation = np.std(data)

    print(f"Range: {range}")
    print(f"Variance: {variance}")
    print(f"Standard Deviation: {standard_deviation}")
```

```python
analyze_data(data)
```

## Q2. Impact of outliers

```python
# adding outliers
for i in range(5):
    data.append(random.randint(200,300))
```

```python
# impact on MOS of outliers
analyze_data(data)
```

## Q3. analyze the impact of transformation x = x *c +d on MOS

```python
import random
import numpy as np
data = [random.randint(1,100) for i in range(1,20)]

print("Before Any Transforamtion")
print(f"data: {data}")
analyze_data(data)

original_data = data

mul = 5
add = 5

transformed_data = [((x * mul) + add) for x in original_data]
print(f"\nTransformed_data: {transformed_data}")
analyze_data(transformed_data)

print("\nFind MOS of transformed data from original data")
print(f"Range: {(max(original_data)-min(original_data))*mul}")
print(f"Variance: {np.var(original_data) * mul**2}")
print(f"Standard Deviation: {np.std(original_data) * abs(mul)}")
```

## Q4. Write a python script that divides the given sample dataset into & quartiles
### [i]   calculate Q1, Q2, Q3
### [ii]  IQR
### [iii] Entries in each quartile

```python
import random

data = [random.randint(1, 100) for i in range(20)]
data.sort()

print("Sample Dataset:")
print(data)

n = len(data)

def median(arr):
    m = len(arr)

    if m % 2 == 0:
        return (arr[m // 2 - 1] + arr[m // 2]) / 2
    else:
        return arr[m // 2]


Q2 = median(data)

if n % 2 == 0:
    lower_half = data[:n // 2]
    upper_half = data[n // 2:]
else:
    lower_half = data[:n // 2]
    upper_half = data[n // 2 + 1:]


Q1 = median(lower_half)
Q3 = median(upper_half)

IQR = Q3 - Q1

print("\nQ1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

Q1_entries = [x for x in data if x <= Q1]
Q2_entries = [x for x in data if Q1 < x <= Q2]
Q3_entries = [x for x in data if Q2 < x <= Q3]
Q4_entries = [x for x in data if x > Q3]

print("\nEntries in each quartile:")

print("First Quartile (Q1):", Q1_entries)
print("Second Quartile (Q2):", Q2_entries)
print("Third Quartile (Q3):", Q3_entries)
print("Fourth Quartile (Q4):", Q4_entries)
```
