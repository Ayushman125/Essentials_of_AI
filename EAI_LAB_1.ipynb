```python
import numpy as np
import pandas as pd
```

1.Evaluate measure of central tendency (MOCT) on Random Samples



Evaluating on Discrete and Continous Array (using Numpy)

```python
#Analysis moct on random samples
np.random.seed(42)
continuous_arr = np.random.normal(loc=50, scale=10, size=100)
discrete_arr = np.random.randint(low=1, high=11, size=20)
print("Discrete Array:", discrete_arr)
```

```python
# --- MEAN ---
np_mean_continous = np.mean(continuous_arr)
np_mean_discrete = np.mean(discrete_arr)

# --- MEDIAN ---
np_median_continous  = np.median(continuous_arr)
np_median_discrete = np.median(discrete_arr)

# --- MODE (Discrete Data) ---
# Count occurrences of unique values

values, counts = np.unique(discrete_arr, return_counts=True)
np_mode_discrete = values[np.argmax(counts)]
raw_mode = pd.Series(continuous_arr).mode().to_numpy()

# Meaningful Continuous Mode: Round to nearest integer or decimal place first
rounded_arr = np.round(continuous_arr, decimals=0)
c_vals, c_counts = np.unique(rounded_arr, return_counts=True)
continuous_mode_rounded = c_vals[c_counts == np.max(c_counts)]



print(f"NumPy Mean   (Continuous): {np_mean_continous:.4f}")
print(f"NumPy Mean   (Discrete): {np_mean_discrete:.4f}")

print(f"NumPy Median (Continuous): {np_median_continous:.4f}")
print(f"NumPy Median (Discrete): {np_median_discrete:.4f}")

print(f"NumPy Mode   (Discrete)  : {np_mode_discrete} (Appeared {np.max(counts)} times)")
print(f"Raw Mode (Unrounded)  : {len(raw_mode)} unique modes (Each appears 1 time)")
print(f"Binned/Rounded Mode   : {continuous_mode_rounded} (Appeared {np.max(c_counts)} times)")
```

Evaluating Across Axes (2D NumPy Arrays) (Using Numpy and Pandas)

```python

# Create a 4x5 array of random numbers
matrix = np.random.randint(1, 100, size=(4, 5))
df = pd.DataFrame(matrix)

print("2D Array / DataFrame:\n", df)
print("-" * 50)

# Column-wise Mean (axis=0)
print("NumPy Column Means :", np.mean(matrix, axis=0))
print("Pandas Column Means:\n", df.mean(axis=0))

# Row-wise Median (axis=1)
print("\nNumPy Row Medians  :", np.median(matrix, axis=1))
print("Pandas Row Medians :\n", df.median(axis=1))
```

2.Introduce Outlier in Samples and analyze the effects of those on MOCT

```python
# 1. Base Clean Sample (Continuous Data)
clean_arr = np.random.normal(loc=50, scale=10, size=100)

# 2. Add Outliers Manually
# We append high extreme values to create a right-skewed dataset
outliers = np.array([500.0, 750.0, 1000.0])
outlier_arr = np.concatenate([clean_arr, outliers])


s_clean = pd.Series(clean_arr)
s_outlier = pd.Series(outlier_arr)
```

Analyzing the Impact of Outliers

```python
def get_central_tendency(series, is_continuous=True):
    mean_val = series.mean()
    median_val = series.median()

    # For continuous data, round to 0 decimals to find a meaningful mode
    data_for_mode = series.round(0) if is_continuous else series
    mode_val = data_for_mode.mode().to_list()

    return mean_val, median_val, mode_val

# Calculate metrics before and after outliers
clean_mean, clean_median, clean_mode = get_central_tendency(s_clean)
out_mean, out_median, out_mode = get_central_tendency(s_outlier)

# Create comparison summary table
comparison_df = pd.DataFrame({
    "Clean Sample |": [clean_mean, clean_median, clean_mode],
    "Sample + Outliers |": [out_mean, out_median, out_mode],
    "Absolute Change": [out_mean - clean_mean, out_median - clean_median, "N/A"]
}, index=["Mean", "Median", "Mode (Rounded)"])

print("=== COMPARISON OF MEASURES OF CENTRAL TENDENCY ===")
print(comparison_df.round(2))
```

3. Create a sample dataset [50 to 100 instances] and perform above mentoned operations on it.

```python
n_instances = 75

# ---------------------------------------------------------
# 1. CREATE CLEAN SAMPLE DATASET (75 Instances)
# ---------------------------------------------------------
clean_data = {
    "model_id": [f"M_{i:03d}" for i in range(1, n_instances + 1)],
    "datasize_kb": np.random.randint(5000, 50000, size=n_instances),
    "accuracy": np.round(np.random.uniform(0.70, 0.98, size=n_instances), 4),
    "training_time_hr": np.round(np.random.uniform(1.0, 12.0, size=n_instances), 2),
    "inference_time_max_sec": np.round(np.random.uniform(0.05, 1.50, size=n_instances), 3)
}

df_clean = pd.DataFrame(clean_data)
print("=== FIRST 5 ROWS OF CLEAN DATASET ===")
print(df_clean.head())
print("-" * 65)

# ---------------------------------------------------------
# 2. HELPER FUNCTION TO COMPUTE CENTRAL TENDENCY
# ---------------------------------------------------------
num_cols = ["datasize_kb", "accuracy", "training_time_hr", "inference_time_max_sec"]

def get_central_tendencies(df, label="Dataset"):
    means = df[num_cols].mean()
    medians = df[num_cols].median()

    # For continuous metrics, mode is calculated after rounding to group nearby values
    modes = df[num_cols].apply(lambda col: col.round(1 if col.name != 'datasize_kb' else -3).mode().iloc[0])

    res = pd.DataFrame({
        f"Mean ({label})": means,
        f"Median ({label})": medians,
        f"Mode ({label})": modes
    })
    return res

# Calculate metrics for clean dataset
ct_clean = get_central_tendencies(df_clean, "Clean")

# ---------------------------------------------------------
# 3. MANUALLY INTRODUCE OUTLIERS
# ---------------------------------------------------------
df_outlier = df_clean.copy()

# Inject extreme synthetic anomalies manually into specific instances
df_outlier.loc[10, "datasize_kb"] = 1_500_000         # Extreme dataset size error
df_outlier.loc[25, "accuracy"] = 0.001                # Failed pipeline accuracy
df_outlier.loc[42, "training_time_hr"] = 450.0        # Hanging GPU training process
df_outlier.loc[60, "inference_time_max_sec"] = 120.0  # Server timeout spike

print("=== INJECTED OUTLIERS ===")
print(df_outlier.loc[[10, 25, 42, 60], ["model_id"] + num_cols])
print("-" * 65)

# ---------------------------------------------------------
# 4. CALCULATE METRICS & ANALYZE OUTLIER EFFECTS
# ---------------------------------------------------------
ct_outlier = get_central_tendencies(df_outlier, "Outlier")

# Combine results for direct side-by-side comparison
comparison = pd.concat([ct_clean, ct_outlier], axis=1)

# Calculate absolute shift percentage in Mean and Median due to outliers
comparison["Mean % Change"] = ((ct_outlier["Mean (Outlier)"] - ct_clean["Mean (Clean)"]) / ct_clean["Mean (Clean)"]) * 100
comparison["Median % Change"] = ((ct_outlier["Median (Outlier)"] - ct_clean["Median (Clean)"]) / ct_clean["Median (Clean)"]) * 100

# Reorder columns for clarity
comparison = comparison[[
    "Mean (Clean)", "Mean (Outlier)", "Mean % Change",
    "Median (Clean)", "Median (Outlier)", "Median % Change",
    "Mode (Clean)", "Mode (Outlier)"
]]

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
print("=== CENTRAL TENDENCY COMPARISON & OUTLIER IMPACT ===")
print(comparison.round(2))
```
